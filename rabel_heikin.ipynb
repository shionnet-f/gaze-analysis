```python
import pandas as pd
```

```python
file_path = "./exported_csv/final_selected_duration.csv" # あなたの環境に合わせて変更
df = pd.read_csv(file_path)
```

```python
df["image_label"] = df["experiment_id"].astype(str) + "-" + (df["trial"] + 1).astype(str)
label_mapping = {
    "不気味": ["1-1", "1-7", "2-5", "2-7", "3-5", "3-8"],
    "中間": ["1-2", "1-8", "2-1", "2-8", "3-2", "3-7"],
    "自然": ["1-4", "1-5", "2-2", "2-4", "3-1", "3-4"],
    "実写真": ["1-3", "1-6", "2-3", "2-6", "3-3", "3-6"],
}
```

```python
def assign_condition(image_label):
    for cond, labels in label_mapping.items():
        if image_label in labels:
            return cond
    return "不明"

df["condition_label"] = df["image_label"].apply(assign_condition)
```

```python
# # count
# aoi_columns = [
#     "left_eye",
#     "right_eye",
#     "nose",
#     "mouth",
#     "parts_aoi_outside",
#     "triangle",
#     "triangle_aoi_outside"
# ]

#duraion
aoi_columns = [
    "left_eye",
    "right_eye",
    "nose",
    "mouth",
    "parts_aoi_outside",
    "triangle",
    "triangle_aoi_outside"
]

print(aoi_columns)
```

```python
# #count
# for aoi in aoi_columns:
#     df[aoi + "_ratio"] = df[aoi] / df["total"]

# duration
for aoi in aoi_columns:
    df[aoi + "_ratio"] = df[aoi] / df["total_duration"]
```

```python
df["subject_id"] = df["subject_id"].astype(str)
```

```python
# 平均値と件数を格納する辞書
aoi_mean_dataframes = {}
aoi_count_dataframes = {}

for aoi in aoi_columns:
    # 平均
    pivot_mean = df.pivot_table(
        index="subject_id",
        columns="condition_label",
        values=aoi + "_ratio",
        aggfunc="mean"
    )
    # 件数
    pivot_count = df.pivot_table(
        index="subject_id",
        columns="condition_label",
        values=aoi + "_ratio",
        aggfunc="count"
    )
    # indexを数値→文字列に揃える
    for pivot in (pivot_mean, pivot_count):
        pivot.index = pivot.index.astype(int)
        pivot.sort_index(inplace=True)
        pivot.index = pivot.index.astype(str)

    # 辞書に格納
    aoi_mean_dataframes[aoi] = pivot_mean
    aoi_count_dataframes[aoi] = pivot_count
```

```python
# CSV保存もする場合
for aoi in aoi_columns:
    mean_filename = f"./exported_csv/imagerabel_AOI_ratio/durations/{aoi}_ratio_by_condition_mean.csv"
    count_filename = f"./exported_csv/imagerabel_AOI_ratio/durations/{aoi}_ratio_by_condition_count.csv"
    aoi_mean_dataframes[aoi].to_csv(mean_filename, encoding="utf-8-sig")
    aoi_count_dataframes[aoi].to_csv(count_filename, encoding="utf-8-sig")
    print(f"保存しました: {mean_filename}, {count_filename}")
```
