```python
import pandas as pd
import os

# ディレクトリ
input_dir = "./exported_csv/fixation_sliced_by_time/"

# 対象範囲
subject_ids = range(1, 20)
experiment_ids = range(1, 4)
trial_nums = range(0, 8)

# 条件マッピング
condition_mapping = {
    "不気味": ["1-1", "1-7", "2-5", "2-7", "3-5", "3-8"],
    "中間": ["1-2", "1-8", "2-1", "2-8", "3-2", "3-7"],
    "自然": ["1-4", "1-5", "2-2", "2-4", "3-1", "3-4"],
    "実写真": ["1-3", "1-6", "2-3", "2-6", "3-3", "3-6"]
}

# 結果を集めるリスト
records = []

for subject_id in subject_ids:
    for experiment_id in experiment_ids:
        for trial_num in trial_nums:
            exp_trial = f"{experiment_id}-{trial_num+1}"

            condition_label = None
            for label, pairs in condition_mapping.items():
                if exp_trial in pairs:
                    condition_label = label
                    break

            if condition_label is None:
                continue

            filename = f"fixation_{subject_id:03}-{experiment_id:03}-{trial_num}_interval_0.0-2.0s.csv"
            file_path = os.path.join(input_dir, filename)

            if not os.path.exists(file_path):
                print(f"スキップ: ファイルが存在しません {file_path}")
                continue

            try:
                df = pd.read_csv(file_path)
                if df.empty:
                    fixation_count = 0
                else:
                    fixation_count = len(df)
            except Exception as e:
                print(f"読み込みエラー: {file_path} ({e})")
                fixation_count = 0

            records.append({
                "subject_id": subject_id,
                "condition": condition_label,
                "trial": trial_num,
                "fixation_count": fixation_count
            })

# データフレーム
counts_df = pd.DataFrame(records)

print("\n=== 全データ ===")
print(counts_df)

# 平均
mean_df = (
    counts_df
    .groupby(["subject_id", "condition"])["fixation_count"]
    .mean()
    .reset_index()
)

# 分母（試行数）
count_df = (
    counts_df
    .groupby(["subject_id", "condition"])["fixation_count"]
    .count()
    .reset_index()
    .rename(columns={"fixation_count": "trial_count"})
)

# ピボット形式
mean_pivot = mean_df.pivot(index="subject_id", columns="condition", values="fixation_count")
count_pivot = count_df.pivot(index="subject_id", columns="condition", values="trial_count")

print("\n=== 平均注視回数 ===")
print(mean_pivot)

print("\n=== 試行数 ===")
print(count_pivot)

# CSV保存
mean_pivot.to_csv("mean_fixation_counts_per_condition.csv", encoding="utf-8-sig")
count_pivot.to_csv("trial_counts_per_condition.csv", encoding="utf-8-sig")
```
