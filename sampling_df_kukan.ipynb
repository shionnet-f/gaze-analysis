```python
import pandas as pd
import os


# === 出力フォルダ（任意）===
output_dir = "./exported_csv/sampling_df_add_kukan"
os.makedirs(output_dir, exist_ok=True)
for subject_id in range(1,20):
    for experiment_id in range(1,4):
        file_path = f"./exported_csv/sampling_df_id{subject_id:03}-{experiment_id:03}.csv"
        print(f"\n処理中: {file_path}")

        # ファイル存在確認
        if not os.path.exists(file_path):
            print("⚠️ ファイルが存在しません。スキップします。")
            continue

        try:
            # データ読み込み
            df = pd.read_csv(file_path)
        except Exception as e:
            print(f"⚠️ 読み込みエラー: {e}")
            continue

        # 空データ確認
        if df.empty:
            print("⚠️ データが空です。スキップします。")
            continue

        # 必要な列があるか確認
        if "start_sec" not in df.columns:
            print("⚠️ 'start_sec' 列がありません。スキップします。")
            continue

        # 新しい列を追加
        df["start_plus_0.5s"] = df["start_sec"] + 0.5
        df["start_plus_2.0s"] = df["start_sec"] + 2.0
        df["start_plus_5.0s"] = df["start_sec"] + 5.0

        # 出力ファイル名
        base_name = os.path.basename(file_path)
        output_path = os.path.join(output_dir, f"with_timepoints_{base_name}")

        # CSV保存
        df.to_csv(output_path, index=False)
        print(f"✅ 保存しました: {output_path}")
```

```python
import pandas as pd
import os

# === 被験者ID・実験課題IDのリスト
subject_ids = [f"{i:03}" for i in range(1, 20)]  # id001 - id019
experiment_ids = [f"{i:03}" for i in range(1, 4)]  # 001 - 003

# === 出力先ディレクトリ
output_dir = "./exported_csv/fixation_counts/IVT/classified_fixations_all"
os.makedirs(output_dir, exist_ok=True)

# === メインループ
for subj_id in subject_ids:
    for exp_id in experiment_ids:
        # sampling_dfファイルパス
        sampling_file = f"./exported_csv/sampling_df_add_kukan/with_timepoints_sampling_df_id{subj_id}-{exp_id}.csv"
        if not os.path.exists(sampling_file):
            print(f"⚠️ samplingファイルが存在しません: {sampling_file}")
            continue
        
        # sampling_df読み込み
        sampling_df = pd.read_csv(sampling_file)

        # fixationファイルパスパターン
        fixation_file_pattern = f"./exported_csv/fixation_counts/IVT/fixations_in_AOI_{subj_id}-{exp_id}-"

        # trialごとに処理
        for trial_num in sampling_df["trial"]:
            fixation_file = f"{fixation_file_pattern}{trial_num}.csv"
            if not os.path.exists(fixation_file):
                print(f"⚠️ fixationファイルが存在しません: {fixation_file}")
                continue

            if os.path.getsize(fixation_file) == 0:
                print(f"⚠️ fixationファイルが0バイトです: {fixation_file}")
                continue

            try:
                fixation_df = pd.read_csv(fixation_file)
            except pd.errors.EmptyDataError:
                print(f"⚠️ EmptyDataError: {fixation_file} が空です。スキップします。")
                continue

            if fixation_df.empty:
                print(f"⚠️ 空データ(行なし): {fixation_file}")
                continue

            # 該当trialのsampling情報
            row_query = sampling_df[sampling_df["trial"] == trial_num]
            if row_query.empty:
                print(f"⚠️ sampling_dfにtrial {trial_num} がありません。スキップします。")
                continue

            row = row_query.iloc[0]
            t0 = row["start_sec"]

            # 区間（相対時間）
            interval_bounds = {
                "0-0.5s": (0, 0.5),
                "0.5-2.0s": (0.5, 2.0),
                "2.0-5.0s": (2.0, 5.0),
                "after_5s": (5.0, float("inf")),
            }

            # 相対時間列を追加
            fixation_df["relative_time"] = fixation_df["start_time"] - t0

            # 区間ごとに分割・保存
            for label, (start_bound, end_bound) in interval_bounds.items():
                subset = fixation_df[
                    (fixation_df["relative_time"] >= start_bound)
                    & (fixation_df["relative_time"] < end_bound)
                ]
                if not subset.empty:
                    out_path = os.path.join(
                        output_dir,
                        f"fixations_id{subj_id}-{exp_id}-trial{trial_num}_{label}.csv"
                    )
                    subset.to_csv(out_path, index=False)
                    print(f"✅ 保存しました: {out_path}")
                else:
                    print(f"⚠️ データなし: id{subj_id}-{exp_id}-trial{trial_num} 区間 {label}")
```
