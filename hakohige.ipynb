```python
import pandas as pd
import matplotlib.pyplot as plt
import japanize_matplotlib

# データ読み込み
df = pd.read_csv("image_hyouka_summary.csv")

# FigureとAxesを用意
fig, ax = plt.subplots(figsize=(10, 6))

# 箱ひげ図を描画（箱の線幅を太くする linewidth=2）
boxprops = dict(linewidth=2)
medianprops = dict(linewidth=2)

df[['実写真', '生成画像(自然)', '生成画像(中間)', '生成画像(不気味)']].boxplot(
    ax=ax,
    boxprops=boxprops,
    medianprops=medianprops
)

# 軸ラベルを大きく
ax.set_ylabel("評価スコア", fontsize=14)
ax.set_xlabel("画像条件", fontsize=14)

# タイトル
# ax.set_title("各条件の評価スコアの分布", fontsize=16)

# 目盛りラベルのフォントサイズ
ax.tick_params(axis='both', labelsize=12)
# Y軸範囲
ax.set_ylim(0, 5.5)

# グリッドオフ
ax.grid(False)


# 表示
plt.show()
```
