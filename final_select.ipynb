```python
# 1. all_ok == True
df_flags_ok = df_flags[df_flags["all_ok"] == True]
```

```python
# 2. 精密度・正確度条件
df_precision_use = df_precision[
    (
        (df_precision["precision_bool"] == 1) & (df_precision["accuracy_bool"] == 1)
    )
    |
    (
        (df_precision["precision_bool"] == 1) & (df_precision["accuracy_bool"] == 0)
    )
]
```

```python
# 3. all_ok と 精密度・正確度条件をマージ
df_ok_and_precision = pd.merge(
    df_flags_ok,
    df_precision_use,
    on=["subject_id", "experiment_id"],
    how="inner"
)
```

```python
# 4. fixation_counts_summary_all_AOI.csv とマージ
df_result = pd.merge(
    df_ok_and_precision,
    df_counts,
    left_on=["subject_id", "experiment_id", "trial"],
    right_on=["id", "exp_id", "trial"],
    how="inner"
)
```

```python
# 出力確認
print(f"抽出結果 {len(df_result)} 行")
df_result
```

```python
# # 必要なら保存
output_path = "./exported_csv/final_selected_count.csv"
df_result.to_csv(output_path, index=False)
print(f"✅ 保存完了: {output_path}")
```
