```python
import pandas as pd
import numpy as np
from statsmodels.stats.anova import AnovaRM
from pingouin import pairwise_ttests
```

```python

# 平均注視割合（先に作った mean.csv）
df_mean = pd.read_csv("./exported_csv/imagerabel_AOI_ratio/triangle_ratio_by_condition_mean.csv")

df_mean.columns = ["subject", "F1", "F2", "F3", "F4"]

df_mean
```

```python
# long形式に変換
df_long = pd.melt(df_mean,
                  id_vars=["subject"],
                  value_vars=["F1", "F2", "F3", "F4"],
                  var_name="Condition",
                  value_name="Score")
```

```python
# 被験者内ANOVA
anova = AnovaRM(data=df_long,
                depvar="Score",
                subject="subject",
                within=["Condition"])
res = anova.fit()
print(res)
```

```python
# 事後検定
posthoc = pairwise_ttests(data=df_long,
                          dv="Score",
                          within="Condition",
                          subject="subject",
                          parametric=True,
                          padjust="bonf")
print(posthoc)
```
